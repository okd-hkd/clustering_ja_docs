```python
import numpy as np
import os
import MeCab

from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# ライセンステキスト以外のファイル名をリストにしていきます。

# os.listdirは指定したディレクトリのファイル・ディレクトリの一覧を返す
sport_names = [_ for _ in os.listdir("livedoor_news/sports-watch") if _ != 'LICENSE.txt']
it_life_hack_names = [_ for _ in os.listdir("livedoor_news/it-life-hack") if _ != 'LICENSE.txt']
movie_enter_names = [_ for _ in os.listdir("livedoor_news/movie-enter") if _ != 'LICENSE.txt']

all_file_names = sport_names + it_life_hack_names + movie_enter_names

def wakati(text):
    mecab = MeCab.Tagger()
    mecab_result = mecab.parse(text).replace("EOS", "").split('\n')
    mecab_result = [i.replace("\t", "_").replace(",","_").split("_") for i in mecab_result if i != ""]
    return mecab_result

def load_live_door_corpus(label, file_name):
    with open("livedoor_news/%s/"%label + file_name, 'r') as f:
        text = f.read()
    # 不要な文字と、コーパスの最初の二行は日付とURLなので削除します。
    text = "".join(text.split('\n')[2:]).replace('\u3000','')
    text = " ".join([w[0] for w in wakati(text)])
    return text
```

```python
# all_file_names
```

```python
with open("all_file_names.txt", 'w') as f:
    for title in all_file_names:
        f.write(title + '\n')
    
```

```python
vectorizer = TfidfVectorizer(use_idf=True, token_pattern=u'(?u)\\b\\w+\\b')

input_data_labels = []
input_data = []

for file_name in all_file_names:
    label = '-'.join(file_name.split('-')[:-1])
    txt = load_live_door_corpus(label, file_name)

    if len(txt) == 0:
        continue

    input_data.append(txt)
    input_data_labels.append(label)

    
# 正規化して変換
vecs = vectorizer.fit_transform(input_data)

# ３クラスに分けるので、n_clusters=3
kmeans_model = KMeans(n_clusters=3, init='k-means++', random_state=10, max_iter=300, tol=1e-04).fit(vecs)

#クラスタリング結果
predict_labels = kmeans_model.labels_
print(predict_labels)
```

```python
# 結果を集計し可視化
res = {0:{},1:{},2:{}}
for p_label, i_label in zip(predict_labels, input_data_labels):
    try:
        res[p_label][i_label] += 1
    except:
        res[p_label][i_label] = 1
print(res)
```
